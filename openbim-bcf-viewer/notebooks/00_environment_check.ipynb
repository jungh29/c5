# 00 Environment Check

Prüft nur die Umgebung und Importe für den V1-Workflow.

```python
import importlib
import sys
from pathlib import Path

def check_import(module_name: str):
    try:
        importlib.import_module(module_name)
        print(f'✅ Import ok: {module_name}')
        return True
    except Exception as exc:
        print(f'❌ Import fehlgeschlagen: {module_name} -> {exc}')
        return False

print(f'Python-Version: {sys.version}')

cwd = Path.cwd()
if (cwd / 'src').exists() and (cwd / 'notebooks').exists():
    project_root = cwd
elif cwd.name == 'notebooks' and (cwd.parent / 'src').exists():
    project_root = cwd.parent
else:
    candidates = [p for p in [cwd, *cwd.parents] if (p / 'src').exists() and (p / 'notebooks').exists()]
    project_root = candidates[0] if candidates else cwd

print(f'Projektroot: {project_root.resolve()}')
print(f"data/: {(project_root / 'data').resolve()} (exists={(project_root / 'data').exists()})")
print(f"exports/: {(project_root / 'exports').resolve()} (exists={(project_root / 'exports').exists()})")

print('\n--- Drittanbieter-Module ---')
check_import('ifcopenshell')
check_import('pandas')
check_import('ipywidgets')
check_import('pythreejs')

print('\n--- Lokale Module ---')
check_import('openbim_viewer.ifc_loader')
check_import('openbim_viewer.bcf_loader')
check_import('openbim_viewer.mapping')
check_import('openbim_viewer.viewer_adapter')
```
